```python
import statsmodels.api as sm
import pandas as pd

portfolio_excess_return_cols = [
    'BV_excess_return', 'BN_excess_return', 'BG_excess_return',
    'SV_excess_return', 'SN_excess_return', 'SG_excess_return'
]

portfolio_betas = {}
time_series_residuals = pd.DataFrame(index=merged_df.index) # DataFrame to store residuals

X_ts = sm.add_constant(merged_df['market_return_excess'])

for portfolio_col in portfolio_excess_return_cols:
    y_ts = merged_df[portfolio_col]
    # Drop rows with NaN values in the current portfolio's excess return
    valid_indices_ts = y_ts.dropna().index
    y_cleaned_ts = y_ts.loc[valid_indices_ts]
    X_cleaned_ts = X_ts.loc[valid_indices_ts]

    model_ts = sm.OLS(y_cleaned_ts, X_cleaned_ts)
    results_ts = model_ts.fit()
    portfolio_betas[portfolio_col] = results_ts.params['market_return_excess']

    # Store residuals, reindexing to the original merged_df index to preserve NaN alignment
    residuals = results_ts.resid
    time_series_residuals[portfolio_col] = pd.Series(residuals, index=valid_indices_ts)

print("Estimated Betas for each portfolio:")
for portfolio, beta in portfolio_betas.items():
    print(f"{portfolio}: {beta:.4f}")
```

```python
import numpy as np

# 1. Calculate the mean vector and covariance matrix of the factor returns
# In this case, the only factor is market excess return.
mu_f = merged_df['market_return_excess'].mean()
# Corrected: Use var() for the variance of a single variable
Sigma_f = merged_df['market_return_excess'].var()

# 2. Calculate the Shanken correction factor
# The formula is (1 + mu_f' * inv(Sigma_f) * mu_f)
# For a single factor, this simplifies to (1 + mu_f^2 / Sigma_f)
shanken_correction_factor = (1 + (mu_f**2) / Sigma_f)

# 3. Multiply the squared standard errors of the Fama-MacBeth lambdas by the Shanken correction factor
lambda_0_shanken_variance = (lambda_0_sem**2) * shanken_correction_factor
lambda_market_shanken_variance = (lambda_market_sem**2) * shanken_correction_factor

# 4. Take the square root of the Shanken-corrected variances to get the Shanken-corrected standard errors
lambda_0_shanken_sem = np.sqrt(lambda_0_shanken_variance)
lambda_market_shanken_sem = np.sqrt(lambda_market_shanken_variance)

# 5. Calculate the t-statistics for the lambdas using the mean lambdas and the Shanken-corrected standard errors
lambda_0_shanken_tstat = lambda_0_mean / lambda_0_shanken_sem
lambda_market_shanken_tstat = lambda_market_mean / lambda_market_shanken_sem

# 6. Print the Shanken-corrected standard errors and t-statistics
print(f"Shanken-corrected Standard Error of lambda_0: {lambda_0_shanken_sem:.4f}")
print(f"Shanken-corrected t-statistic for lambda_0: {lambda_0_shanken_tstat:.4f}")
print(f"Shanken-corrected Standard Error of lambda_market: {lambda_market_shanken_sem:.4f}")
print(f"Shanken-corrected t-statistic for lambda_market: {lambda_market_shanken_tstat:.4f}")
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from scipy.stats import chi2

# Ensure time_series_residuals and pricing_errors are aligned and complete
# The BV_excess_return portfolio has some NaN values in its time-series residuals.
# To calculate the covariance matrix, we will drop rows where any residual is NaN.
# This means we will only use time periods where all portfolios have valid residuals.
# However, dropping rows with any NaN might remove too much data. A common practice
# is to use pairwise deletion for covariance, but for matrix inversion, it's better
# to have a consistent set of observations. Given the BV_excess_return column
# has 115 non-nulls out of 132, dropping rows with any NaN will result in 115 valid observations.

# Let's align the residuals and pricing errors.
# Identify portfolios that are present in both `time_series_residuals` and `pricing_errors`
common_portfolios = list(set(time_series_residuals.columns) & set(pricing_errors.index))

# Filter `time_series_residuals` to only include `common_portfolios`
cleaned_time_series_residuals = time_series_residuals[common_portfolios]

# Drop rows from `cleaned_time_series_residuals` that have *any* NaN across these common portfolios
# This ensures a consistent set of time periods for the covariance matrix calculation.
cleaned_time_series_residuals_no_nan = cleaned_time_series_residuals.dropna(how='any')

# Filter `pricing_errors` to only include `common_portfolios`
pricing_errors_for_chi2 = pricing_errors.loc[common_portfolios]

# Check if enough data remains after dropping NaNs
if len(cleaned_time_series_residuals_no_nan) < 2:
    print("Not enough valid time series observations after dropping NaNs to calculate covariance matrix.")
    chi2_statistic = np.nan
    df_chi2 = np.nan
    p_value = np.nan
else:
    # 1. Calculate the covariance matrix of the cleaned time-series residuals (S_e)
    # This is an N x N matrix where N is the number of portfolios being tested.
    S_e = cleaned_time_series_residuals_no_nan.cov()

    # Number of time periods (T) used for the residuals (after cleaning NaNs)
    T_actual = len(cleaned_time_series_residuals_no_nan)

    # Number of factors (K) - 1 for market factor in the second pass
    K = 1

    # Number of portfolios (N) being tested
    N_test = len(common_portfolios)

    # Ensure S_e is invertible
    try:
        # Invert the covariance matrix
        inv_S_e = pd.DataFrame(np.linalg.inv(S_e.values), columns=S_e.columns, index=S_e.index)

        # 2. Calculate the Chi-square statistic (Wald test) for pricing errors
        # Wald = T_actual * (pricing_errors_vector)' * inv(S_e) * (pricing_errors_vector)
        chi2_statistic = T_actual * pricing_errors_for_chi2.values.T @ inv_S_e.values @ pricing_errors_for_chi2.values

        # 3. Degrees of freedom for the test
        # For a Wald test on N pricing errors, with K factors and an intercept (lambda_0) in the cross-sectional regression,
        # the degrees of freedom are N - (number of parameters estimated in CS regression) = N - (K + 1)
        df_chi2 = N_test - K - 1

        if df_chi2 <= 0:
            print(f"Degrees of freedom ({df_chi2}) is not positive. Cannot perform Chi-square test meaningfully.")
            p_value = np.nan
        else:
            # 4. Calculate the p-value
            p_value = 1 - chi2.cdf(chi2_statistic, df_chi2)

    except np.linalg.LinAlgError:
        print("Covariance matrix of time-series residuals is singular, cannot invert. Cannot perform Chi-square test.")
        chi2_statistic = np.nan
        df_chi2 = np.nan
        p_value = np.nan

print(f"Chi-square statistic (Wald test) for pricing errors: {chi2_statistic:.4f}")
print(f"Degrees of freedom: {df_chi2}")
print(f"P-value: {p_value:.4f}")

print("\nInterpretation:")
print("The null hypothesis of the Wald test is that all pricing errors (alphas) are jointly zero.")
print("A small p-value (typically < 0.05) would lead to the rejection of the null hypothesis,")
print("suggesting that the model does not fully explain the cross-section of expected returns.")
```

### Interpretation of Pricing Errors

The **Root Mean Squared Pricing Error (RMSE)** provides a single, aggregate measure of how well the model prices the portfolios. A smaller RMSE indicates a better fit, meaning the model's expected returns are closer to the actual mean returns.

The **Sum of Squared Pricing Errors (SSPE)** is another aggregate measure, emphasizing larger individual pricing errors. While both provide insights into the model's overall fit, interpreting whether these values are 'good' or 'bad' often requires comparison to other models or benchmarks.

**Note on Chi-square statistic**: A formal Chi-square (or GMM J-test) for pricing errors in a Fama-MacBeth framework requires the full covariance matrix of the pricing errors, which is derived from the GMM estimation process. The current implementation of Fama-MacBeth does not directly provide this. Therefore, a rigorous Chi-square test with a p-value cannot be performed with the current output. The Root Mean Squared Pricing Error (RMSE) and Sum of Squared Pricing Errors (SSPE) serve as practical aggregate measures of the model's ability to explain the cross-section of returns.
