## Model

```python
import statsmodels.api as sm
import numpy as np

# 1. Define a list containing the names of the portfolio excess return columns.
portfolio_excess_return_cols = [
    'BV_excess_return', 'BN_excess_return', 'BG_excess_return',
    'SV_excess_return', 'SN_excess_return', 'SG_excess_return'
]

# 2. Initialize an empty dictionary to store the estimated betas for each portfolio.
portfolio_betas_ccapm_cay = {}

# 3. Define the independent variables for the time-series regression as 'consumption_growth', 'consumption_growth_x_cay_standardized' and 'cay_standardized'
X_ts = merged_df[['consumption_growth', 'consumption_growth_x_cay_standardized', 'cay_standardized']].copy()

# 4. Add a constant term to the independent variables DataFrame using sm.add_constant.
X_ts = sm.add_constant(X_ts)

# 5. Iterate through each portfolio excess return column:
print("Performing time-series regressions for CCAPM with consumption growth, consumption growth x cay interaction, and cay...")
for portfolio_col in portfolio_excess_return_cols:
    # a. Define the dependent variable as the current portfolio's excess return.
    y_ts = merged_df[portfolio_col]

    # b. Drop rows with NaN values in the dependent variable and the corresponding rows in the independent variables DataFrame.
    valid_indices = y_ts.dropna().index
    y_ts_cleaned = y_ts.loc[valid_indices]
    X_ts_cleaned = X_ts.loc[valid_indices]

    # Drop rows with NaN values in X_ts_cleaned before fitting the model
    valid_X_indices = X_ts_cleaned.dropna().index
    y_ts_cleaned = y_ts_cleaned.loc[valid_X_indices]
    X_ts_cleaned = X_ts_cleaned.loc[valid_X_indices]


    if not X_ts_cleaned.empty and not y_ts_cleaned.empty:
        # c. Fit an OLS model using sm.OLS with the cleaned dependent and independent variables.
        model_ts = sm.OLS(y_ts_cleaned, X_ts_cleaned)
        results_ts = model_ts.fit()

        # d. Store the estimated coefficients for 'consumption_growth', 'consumption_growth_x_cay_standardized' and 'cay_standardized'
        portfolio_betas_ccapm_cay[portfolio_col] = {
            'beta_consumption': results_ts.params['consumption_growth'],
            'beta_consumption_x_cay': results_ts.params['consumption_growth_x_cay_standardized'],
            'beta_cay': results_ts.params['cay_standardized']
        }
    else:
         # Handle cases where there are no valid data points for regression
        portfolio_betas_ccapm_cay[portfolio_col] = {
            'beta_consumption': np.nan,
            'beta_consumption_x_cay': np.nan,
            'beta_cay': np.nan
        }

# 6. Print the estimated betas for each portfolio.
print("Time-series regressions complete. Estimated Betas for CCAPM with consumption growth, consumption growth x cay interaction, and cay:")
for portfolio, betas in portfolio_betas_ccapm_cay.items():
    print(f"{portfolio}:")
    print(f"  Beta Consumption: {betas['beta_consumption']:.4f}")
    print(f"  Beta Consumption x CAY: {betas['beta_consumption_x_cay']:.4f}")
    print(f"  Beta CAY: {betas['beta_cay']:.4f}")
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from scipy.stats import chi2

# Create a DataFrame of the estimated betas for cross-sectional regression
betas_df_ccapm_cay = pd.DataFrame.from_dict(portfolio_betas_ccapm_cay, orient='index')
betas_df_ccapm_cay = sm.add_constant(betas_df_ccapm_cay) # Add constant for lambda_0

# Perform cross-sectional regressions
print("\nPerforming cross-sectional regressions for CCAPM with consumption growth, consumption growth x cay interaction, and cay...")
lambda_estimates_ccapm_cay = []
all_cs_residuals = [] # List to store cross-sectional residuals for each time period

portfolio_excess_return_cols = [
    'BV_excess_return', 'BN_excess_return', 'BG_excess_return',
    'SV_excess_return', 'SN_excess_return', 'SG_excess_return'
]

for index, row in merged_df.iterrows():
    y_cs = row[portfolio_excess_return_cols]

    # Drop NaN values from y_cs and get the corresponding betas from betas_df_ccapm_cay
    valid_portfolios = y_cs.dropna().index
    y_cs_cleaned = y_cs.loc[valid_portfolios].astype(float)

    # Select the betas corresponding to the valid portfolios
    betas_for_cs = betas_df_ccapm_cay.loc[valid_portfolios].astype(float)

    # Ensure there are enough valid data points and betas for regression (at least 4 for intercept and three betas)
    if len(valid_portfolios) >= 4 and not betas_for_cs.empty:
        try:
            model_cs = sm.OLS(y_cs_cleaned, betas_for_cs)
            results_cs = model_cs.fit()
            # Ensure the number of parameters is 4 before appending (lambda_0, lambda_consumption, lambda_consumption_x_cay, lambda_cay)
            if len(results_cs.params) == 4:
                lambda_estimates_ccapm_cay.append(results_cs.params.tolist())
                # Store residuals for chi-square calculation
                residuals_series = results_cs.resid.rename(row.name)
                all_cs_residuals.append(residuals_series)
            else:
                 # Append NaNs if the number of parameters is not as expected
                lambda_estimates_ccapm_cay.append([np.nan, np.nan, np.nan, np.nan])
        except Exception as e:
            # Handle potential errors during regression
            print(f"Regression error at index {index}: {e}")
            lambda_estimates_ccapm_cay.append([np.nan, np.nan, np.nan, np.nan])
    else:
        # Append NaNs if regression cannot be performed
        lambda_estimates_ccapm_cay.append([np.nan, np.nan, np.nan, np.nan])

lambda_df_ccapm_cay = pd.DataFrame(lambda_estimates_ccapm_cay, columns=['lambda_0', 'lambda_consumption', 'lambda_consumption_x_cay', 'lambda_cay'])
print("Cross-sectional regressions complete.")

# Create DataFrame of all cross-sectional residuals
df_all_cs_residuals = pd.DataFrame(all_cs_residuals)

# 3. Compute Fama–MacBeth lambdas and standard errors
print("\nComputing Fama-MacBeth lambdas and standard errors for CCAPM with consumption growth, consumption growth x cay interaction, and cay...")
lambda_0_mean_ccapm_cay = lambda_df_ccapm_cay['lambda_0'].mean()
lambda_0_sem_ccapm_cay = lambda_df_ccapm_cay['lambda_0'].sem()
lambda_consumption_mean_ccapm_cay = lambda_df_ccapm_cay['lambda_consumption'].mean()
lambda_consumption_sem_ccapm_cay = lambda_df_ccapm_cay['lambda_consumption'].sem()
lambda_consumption_x_cay_mean_ccapm_cay = lambda_df_ccapm_cay['lambda_consumption_x_cay'].mean()
lambda_consumption_x_cay_sem_ccapm_cay = lambda_df_ccapm_cay['lambda_consumption_x_cay'].sem()
lambda_cay_mean_ccapm_cay = lambda_df_ccapm_cay['lambda_cay'].mean()
lambda_cay_sem_ccapm_cay = lambda_df_ccapm_cay['lambda_cay'].sem()


print("Fama-MacBeth Lambda Estimates (CCAPM with consumption growth, consumption growth x cay interaction, and cay):")
print(f"Mean of lambda_0: {lambda_0_mean_ccapm_cay:.4f}")
print(f"Standard Error of lambda_0: {lambda_0_sem_ccapm_cay:.4f}")
print(f"Mean of lambda_consumption: {lambda_consumption_mean_ccapm_cay:.4f}")
print(f"Standard Error of lambda_consumption: {lambda_consumption_sem_ccapm_cay:.4f}")
print(f"Mean of lambda_consumption_x_cay: {lambda_consumption_x_cay_mean_ccapm_cay:.4f}")
print(f"Standard Error of lambda_consumption_x_cay: {lambda_consumption_x_cay_sem_ccapm_cay:.4f}")
print(f"Mean of lambda_cay: {lambda_cay_mean_ccapm_cay:.4f}")
print(f"Standard Error of lambda_cay: {lambda_cay_sem_ccapm_cay:.4f}")


# 4. Apply Shanken correction
print("\nApplying Shanken correction...")
# 1. Calculate the mean vector and covariance matrix of the factors
# Factors are consumption growth, consumption_growth_x_cay_standardized, and cay_standardized
factors_ccapm_cay_shanken = merged_df[['consumption_growth', 'consumption_growth_x_cay_standardized', 'cay_standardized']].dropna()
mu_f_ccapm_cay = factors_ccapm_cay_shanken.mean().values
Sigma_f_ccapm_cay = factors_ccapm_cay_shanken.cov().values

# 2. Calculate the Shanken correction factor
# Formula: (1 + mu_f' * inv(Sigma_f) * mu_f)
try:
    Sigma_f_inv_ccapm_cay = np.linalg.inv(Sigma_f_ccapm_cay)
    shanken_correction_factor_ccapm_cay = (1 + mu_f_ccapm_cay.T @ Sigma_f_inv_ccapm_cay @ mu_f_ccapm_cay)
except np.linalg.LinAlgError:
    print("Could not compute inverse of the covariance matrix. Skipping Shanken correction.")
    shanken_correction_factor_ccapm_cay = np.nan


if not np.isnan(shanken_correction_factor_ccapm_cay):
    # 3. Multiply the squared standard errors of the Fama-MacBeth lambdas by the Shanken correction factor
    lambda_0_shanken_variance_ccapm_cay = (lambda_0_sem_ccapm_cay**2) * shanken_correction_factor_ccapm_cay
    lambda_consumption_shanken_variance_ccapm_cay = (lambda_consumption_sem_ccapm_cay**2) * shanken_correction_factor_ccapm_cay
    lambda_consumption_x_cay_shanken_variance_ccapm_cay = (lambda_consumption_x_cay_sem_ccapm_cay**2) * shanken_correction_factor_ccapm_cay
    lambda_cay_shanken_variance_ccapm_cay = (lambda_cay_sem_ccapm_cay**2) * shanken_correction_factor_ccapm_cay


    # 4. Take the square root of the Shanken-corrected variances to get the Shanken-corrected standard errors
    lambda_0_shanken_sem_ccapm_cay = np.sqrt(lambda_0_shanken_variance_ccapm_cay)
    lambda_consumption_shanken_sem_ccapm_cay = np.sqrt(lambda_consumption_shanken_variance_ccapm_cay)
    lambda_consumption_x_cay_shanken_sem_ccapm_cay = np.sqrt(lambda_consumption_x_cay_shanken_variance_ccapm_cay)
    lambda_cay_shanken_sem_ccapm_cay = np.sqrt(lambda_cay_shanken_variance_ccapm_cay)


    # 5. Calculate the t-statistics using the mean lambdas and Shanken-corrected standard errors
    lambda_0_shanken_tstat_ccapm_cay = lambda_0_mean_ccapm_cay / lambda_0_shanken_sem_ccapm_cay
    lambda_consumption_shanken_tstat_ccapm_cay = lambda_consumption_mean_ccapm_cay / lambda_consumption_shanken_sem_ccapm_cay
    lambda_consumption_x_cay_shanken_tstat_ccapm_cay = lambda_consumption_x_cay_mean_ccapm_cay / lambda_consumption_x_cay_shanken_sem_ccapm_cay
    lambda_cay_shanken_tstat_ccapm_cay = lambda_cay_mean_ccapm_cay / lambda_cay_shanken_sem_ccapm_cay


    # 6. Print the Shanken-corrected standard errors and t-statistics
    print("\nShanken-corrected Results (CCAPM with consumption growth, consumption growth x cay interaction, and cay):")
    print(f"Shanken-corrected Standard Error of lambda_0: {lambda_0_shanken_sem_ccapm_cay:.4f}")
    print(f"Shanken-corrected t-statistic for lambda_0: {lambda_0_shanken_tstat_ccapm_cay:.4f}")
    print(f"Shanken-corrected Standard Error of lambda_consumption: {lambda_consumption_shanken_sem_ccapm_cay:.4f}")
    print(f"Shanken-corrected t-statistic for lambda_consumption: {lambda_consumption_shanken_tstat_ccapm_cay:.4f}")
    print(f"Shanken-corrected Standard Error of lambda_consumption_x_cay: {lambda_consumption_x_cay_shanken_sem_ccapm_cay:.4f}")
    print(f"Shanken-corrected t-statistic for lambda_consumption_x_cay: {lambda_consumption_x_cay_shanken_tstat_ccapm_cay:.4f}")
    print(f"Shanken-corrected Standard Error of lambda_cay: {lambda_cay_shanken_sem_ccapm_cay:.4f}")
    print(f"Shanken-corrected t-statistic for lambda_cay: {lambda_cay_shanken_tstat_ccapm_cay:.4f}")

else:
    print("\nShanken correction could not be applied.")

# 5. Compute cross-sectional R²
print("\nComputing average cross-sectional R-squared...")
cs_r_squared_values_ccapm_cay = []

for index, row in merged_df.iterrows():
    y_cs = row[portfolio_excess_return_cols]

    valid_portfolios = y_cs.dropna().index
    y_cs_cleaned = y_cs.loc[valid_portfolios].astype(float)

    betas_for_cs = betas_df_ccapm_cay.loc[valid_portfolios].astype(float)

    if len(valid_portfolios) >= 4 and not betas_for_cs.empty:
        try:
            model_cs = sm.OLS(y_cs_cleaned, betas_for_cs)
            results_cs = model_cs.fit()
            if len(results_cs.params) == 4:
                 # Calculate R-squared for the cross-sectional regression
                y_cs_mean = y_cs_cleaned.mean()
                ss_total = ((y_cs_cleaned - y_cs_mean) ** 2).sum()
                # Calculate predicted values using the cross-sectional regression results
                predicted_y_cs = results_cs.predict(betas_for_cs)
                ss_residual = ((y_cs_cleaned - predicted_y_cs) ** 2).sum()
                # Avoid division by zero if ss_total is zero
                if ss_total > 0:
                    cs_r_squared = 1 - (ss_residual / ss_total)
                    cs_r_squared_values_ccapm_cay.append(cs_r_squared)
                else:
                    cs_r_squared_values_ccapm_cay.append(np.nan) # Append NaN if ss_total is zero
            else:
                cs_r_squared_values_ccapm_cay.append(np.nan)
        except Exception as e:
            print(f"Regression error at index {index}: {e}")
            cs_r_squared_values_ccapm_cay.append(np.nan)
    else:
        cs_r_squared_values_ccapm_cay.append(np.nan)


average_cs_r_squared_ccapm_cay = np.nanmean(cs_r_squared_values_ccapm_cay)

print(f"Average Cross-Sectional R-squared (CCAPM with consumption growth, consumption growth x cay interaction, and cay): {average_cs_r_squared_ccapm_cay:.4f}")


# 6. Report results (Summary)
print("\n--- CCAPM with consumption growth, consumption growth x cay interaction, and cay Model Results ---")
print(f"Mean lambda_consumption: {lambda_consumption_mean_ccapm_cay:.4f}")
if not np.isnan(shanken_correction_factor_ccapm_cay):
    print(f"Shanken-corrected t-statistic for lambda_consumption: {lambda_consumption_shanken_tstat_ccapm_cay:.4f}")
else:
     print("Shanken-corrected t-statistic for lambda_consumption: N/A (Shanken correction not applied)")

print(f"Mean lambda_consumption_x_cay: {lambda_consumption_x_cay_mean_ccapm_cay:.4f}")
if not np.isnan(shanken_correction_factor_ccapm_cay):
    print(f"Shanken-corrected t-statistic for lambda_consumption_x_cay: {lambda_consumption_x_cay_shanken_tstat_ccapm_cay:.4f}")
else:
    print("Shanken-corrected t-statistic for lambda_consumption_x_cay: N/A (Shanken correction not applied)")

print(f"Mean lambda_cay: {lambda_cay_mean_ccapm_cay:.4f}")
if not np.isnan(shanken_correction_factor_ccapm_cay):
    print(f"Shanken-corrected t-statistic for lambda_cay: {lambda_cay_shanken_tstat_ccapm_cay:.4f}")
else:
    print("Shanken-corrected t-statistic for lambda_cay: N/A (Shanken correction not applied)")

print(f"Mean lambda_0: {lambda_0_mean_ccapm_cay:.4f}")
if not np.isnan(shanken_correction_factor_ccapm_cay):
    print(f"Shanken-corrected t-statistic for lambda_0: {lambda_0_shanken_tstat_ccapm_cay:.4f}")
else:
    print("Shanken-corrected t-statistic for lambda_0: N/A (Shanken correction not applied)")


print(f"Average cross-sectional R-squared: {average_cs_r_squared_ccapm_cay:.4f}")

print("\n--- CCAPM with consumption growth, consumption growth x cay interaction, and cay Summary ---")
print("Key Findings:")
print(f"* The estimated betas for the six portfolios show varying sensitivities to consumption growth, the consumption growth x cay interaction term, and cay.")
print(f"* The mean estimated consumption risk premium (lambda_consumption) is {lambda_consumption_mean_ccapm_cay:.4f}.")
print(f"* The mean estimated consumption growth x cay risk premium (lambda_consumption_x_cay) is {lambda_consumption_x_cay_mean_ccapm_cay:.4f}.")
print(f"* The mean estimated cay risk premium (lambda_cay) is {lambda_cay_mean_ccapm_cay:.4f}.")
print(f"* The mean estimated zero-beta rate (lambda_0) is {lambda_0_mean_ccapm_cay:.4f}.")

if not np.isnan(shanken_correction_factor_ccapm_cay):
    print("* Shanken-corrected t-statistics indicate the statistical significance of the estimated risk premia.")
    print(f"  - Consumption Risk Premium t-statistic: {lambda_consumption_shanken_tstat_ccapm_cay:.4f}")
    print(f"  - Consumption x CAY Risk Premium t-statistic: {lambda_consumption_x_cay_shanken_tstat_ccapm_cay:.4f}")
    print(f"  - CAY Risk Premium t-statistic: {lambda_cay_shanken_tstat_ccapm_cay:.4f}")
    print(f"  - Zero-Beta Rate t-statistic: {lambda_0_shanken_tstat_ccapm_cay:.4f}")
else:
    print("* Shanken correction could not be applied due to issues with the covariance matrix.")

print(f"* The average cross-sectional R-squared of {average_cs_r_squared_ccapm_cay:.4f} suggests the proportion of the cross-sectional variation in portfolio excess returns explained by the model.")

print("\nNext Steps:")
print("- Summarize all model results in a comprehensive table.")
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from scipy.stats import chi2

# 1. Get the estimated lambdas (mean risk premia) from the Fama-MacBeth second-pass regression
lambda_0 = lambda_0_mean_ccapm_cay
lambda_consumption = lambda_consumption_mean_ccapm_cay
lambda_consumption_x_cay = lambda_consumption_x_cay_mean_ccapm_cay
lambda_cay = lambda_cay_mean_ccapm_cay

# 2. Get the estimated betas for each portfolio from the first-pass time-series regressions
#    This is already stored in the 'portfolio_betas_ccapm_cay' dictionary

# 3. Calculate the Mean Pricing Error (MPE) for each portfolio
mean_pricing_errors = {}
portfolio_excess_return_cols = [
    'BV_excess_return', 'BN_excess_return', 'BG_excess_return',
    'SV_excess_return', 'SN_excess_return', 'SG_excess_return'
]

for portfolio_col in portfolio_excess_return_cols:
    # a. Get the betas for the current portfolio
    betas = portfolio_betas_ccapm_cay.get(portfolio_col, {})
    beta_consumption = betas.get('beta_consumption', np.nan)
    beta_consumption_x_cay = betas.get('beta_consumption_x_cay', np.nan)
    beta_cay = betas.get('beta_cay', np.nan)

    # Only proceed if all necessary betas are available
    if not np.isnan(beta_consumption) and not np.isnan(beta_consumption_x_cay) and not np.isnan(beta_cay):
        # b. Calculate the predicted excess return
        predicted_excess_return = lambda_0 + \
                                  beta_consumption * lambda_consumption + \
                                  beta_consumption_x_cay * lambda_consumption_x_cay + \
                                  beta_cay * lambda_cay

        # c. Calculate the average realized excess return for the portfolio
        #    Need to drop NaNs before calculating the mean, as BV_excess_return had NaNs
        average_realized_excess_return = merged_df[portfolio_col].mean()

        # d. Calculate the pricing error
        pricing_error = average_realized_excess_return - predicted_excess_return
        mean_pricing_errors[portfolio_col] = pricing_error
    else:
        mean_pricing_errors[portfolio_col] = np.nan

# 4. Display the Mean Pricing Errors
print("\nMean Pricing Errors for each portfolio:")
for portfolio, mpe in mean_pricing_errors.items():
    print(f"  {portfolio}: {mpe:.4f}")

# Create a DataFrame for better display of individual MPEs
df_mpe = pd.DataFrame(mean_pricing_errors.items(), columns=['Portfolio', 'Mean Pricing Error'])
df_mpe.set_index('Portfolio', inplace=True)
print("\nMean Pricing Errors (DataFrame):")
display(df_mpe.round(4))

# 5. Calculate Overall Mean Pricing Error
overall_mean_pricing_error = df_mpe['Mean Pricing Error'].mean()
print(f"\nOverall Mean Pricing Error across all portfolios: {overall_mean_pricing_error:.4f}")

# 6. Calculate Chi-square statistic for the errors (GRS-like test for joint significance of alphas)
# This uses the time-series of cross-sectional residuals collected in df_all_cs_residuals
if 'df_all_cs_residuals' in globals() and not df_all_cs_residuals.empty:
    try:
        # Ensure df_all_cs_residuals is aligned with the portfolios for which MPEs are calculated
        # Filter residuals to include only portfolios with calculated MPEs
        valid_portfolios_in_mpe = [p for p, mpe in mean_pricing_errors.items() if not np.isnan(mpe)]
        df_all_cs_residuals_filtered = df_all_cs_residuals[valid_portfolios_in_mpe].dropna()

        if not df_all_cs_residuals_filtered.empty:
            # Covariance matrix of the cross-sectional residuals
            cov_epsilon = df_all_cs_residuals_filtered.cov()

            # Vector of mean pricing errors for valid portfolios
            alpha_vec = df_mpe.loc[valid_portfolios_in_mpe, 'Mean Pricing Error'].values

            # Number of time periods for which residuals are available
            T = len(df_all_cs_residuals_filtered)

            # Number of portfolios included in the test
            N_portfolios = len(alpha_vec)

            # Check for singularity before inverting
            if np.linalg.det(cov_epsilon) == 0:
                print("\nCovariance matrix of residuals is singular. Cannot compute Chi-square statistic.")
            else:
                inv_cov_epsilon = np.linalg.inv(cov_epsilon)
                chi_square_stat = T * (alpha_vec.T @ inv_cov_epsilon @ alpha_vec)
                df_chi2 = N_portfolios  # Degrees of freedom is the number of portfolios
                p_value_chi2 = 1 - chi2.cdf(chi_square_stat, df_chi2)

                print(f"\nChi-square statistic for joint significance of pricing errors: {chi_square_stat:.4f}")
                print(f"Degrees of Freedom: {df_chi2}")
                print(f"P-value: {p_value_chi2:.4f}")

                if p_value_chi2 < 0.05:
                    print("Conclusion: Reject the null hypothesis that all pricing errors are jointly zero (at 5% significance level).")
                else:
                    print("Conclusion: Fail to reject the null hypothesis that all pricing errors are jointly zero (at 5% significance level).")
        else:
            print("\nNot enough valid cross-sectional residuals to compute Chi-square statistic.")
    except Exception as e:
        print(f"\nError computing Chi-square statistic: {e}")
else:
    print("\nCross-sectional residuals (df_all_cs_residuals) not found or empty. Cannot compute Chi-square statistic.")
```
